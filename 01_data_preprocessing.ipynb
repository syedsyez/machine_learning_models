```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("data/raw/data.csv")

print("Dataset shape:", df.shape)

df.head()
print(df.columns.tolist())
df.info()
df.head()
df.describe()
print(df["diagnosis"].value_counts())
```

```python
df.drop(
    columns=["id", "Unnamed: 32"],
    errors="ignore",
    inplace=True
)

print("Shape after removing unnecessary columns:", df.shape)
print("Duplicate records:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
print(df.isnull().sum().sum())
```

```python
df["diagnosis"] = df["diagnosis"].map({
    "M": 1,
    "B": 0
})

print(df["diagnosis"].value_counts())
X = df.drop("diagnosis", axis=1)

y = df["diagnosis"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
```

```python
os.makedirs("data/processed", exist_ok=True)
```

```python
train_data = X_train.copy()
train_data["diagnosis"] = y_train

train_data.to_csv(
    "data/processed/train_data.csv",
    index=False
)

test_data = X_test.copy()
test_data["diagnosis"] = y_test

test_data.to_csv(
    "data/processed/test_data.csv",
    index=False
)
```

```python
print("Training dataset:", train_data.shape)
print("Testing dataset:", test_data.shape)

print("\nTraining class distribution:")
print(train_data["diagnosis"].value_counts())

print("\nTesting class distribution:")
print(test_data["diagnosis"].value_counts())

print("\nMissing values:")
print(df.isnull().sum().sum())
```
